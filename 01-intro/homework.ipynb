```python
# Question 6
horsepower_col = df['horsepower']
```

```python
initial_median_horsepower = horsepower_col.median()
initial_median_horsepower
```

```python
most_frequent = horsepower_col.value_counts().idxmax()
most_frequent
```

```python
after_fill_median = horsepower_col.fillna(most_frequent).median()
```

```python
(initial_median_horsepower, after_fill_median)
```

```python
initial_median_horsepower > after_fill_median
```

```python
# Question 7
asia_cars_df = df[df['origin'] == 'Asia'][['vehicle_weight', 'model_year']].head(7)
asia_cars_df
```

```python
X = asia_cars_df.to_numpy()
X
```

```python
XTX = X.T.dot(X)
XTX
```

```python
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
y
```

```python
w = (XTX_inv.dot(X.T)).dot(y)
w
```
